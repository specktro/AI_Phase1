## Exercise 1 - Basic operations
Given the following information:
``` python
a = [2, -1, 4]
b = [3,  2, -1]
c = -2
```
Make the following operations:
1. a + b
2. c * a
3. a - b
4. a · b

````python
import math

import numpy as np

# Information
a = np.array([2, -1, 4])
b = np.array([3, 2, -1])
c = -2
````

````python
vector_one = a + b
print(vector_one.__str__())
````

````python
vector_two = c * a
print(vector_two.__str__())
````

````python
vector_three = a - b
print(vector_three.__str__())
````

````python
dot = np.dot(a, b)
print(dot)
````

## Exercise 2 - Norm L2
Given the following vector `v = [1, 2, 2]` let's calculate the following operations:
1. Calculate ‖v‖₂
2. Calculate the unit vector v̂ = v / ‖v‖
3. Verify ‖v̂‖ = 1.0 — why?

````python
v = np.array([1, 2, 2])
````

````python
norm_l2 = np.linalg.norm(v)
print(norm_l2)
````

````python
unit_v = v / norm_l2
print(unit_v)
print(np.linalg.norm(unit_v))
````

## Exercise 3 - Norm identity/Dot product
Given the following vector `v = [3, -4, 0, 1]` let's calculate the following operations:
1. Calculate v · v
2. Calculate ‖v‖²

````python
v = np.array([3, -4, 0, 1])
````

````python
new_dot_product = v @ v
print(new_dot_product)
````

````python
dot_product = np.linalg.norm(v) ** 2
print(dot_product)
````

## Exercise 4 - Orthogonality
With the following vector we are going to calculate the orthogonality between them:
- u = [1, 2, 3]
- v = [-3, 0, 1]
- w = [2, -5, 4]

So, calculate the following operations `u·v`, `u·w` and `v·w` and determine which of them are orthogonal

````python
u = np.array([1, 2, 3])
v = np.array([-3, 0, 1])
w = np.array([2, -5, 4])
````

````python
u_v = u @ v
print(u_v) # 0 -> this are orthogonal
````

````python
u_w = u @ w
print(u_w) # > 0 acute angle
````

````python
v_w = v @ w
print(v_w) # < 0 obtuse angle
````

````python
angle = math.degrees(math.acos(u_v / np.linalg.norm(u) * np.linalg.norm(v)))
print(angle)
````
