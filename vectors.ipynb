## Exercise 1 - Basic operations
Given the following information:
``` python
a = [2, -1, 4]
b = [3,  2, -1]
c = -2
```
Make the following operations:
1. a + b
2. c * a
3. a - b
4. a · b

````python
import numpy as np

# Information
a = np.array([2, -1, 4])
b = np.array([3, 2, -1])
c = -2
````

````python
vector_one = a + b
print(vector_one.__str__())
````

````python
vector_two = c * a
print(vector_two.__str__())
````

````python
vector_three = a - b
print(vector_three.__str__())
````

````python
dot = np.dot(a, b)
print(dot)
````

## Exercise 2 - Norm L2
Given the following vector `v = [1, 2, 2]` let's calculate the following operations:
1. Calculate ‖v‖₂
2. Calculate the unit vector v̂ = v / ‖v‖
3. Verify ‖v̂‖ = 1.0 — why?

````python
v = np.array([1, 2, 2])
````

````python
norm_l2 = np.linalg.norm(v)
print(norm_l2)
````

````python
unit_v = v / norm_l2
print(unit_v)
print(np.linalg.norm(unit_v))
````
