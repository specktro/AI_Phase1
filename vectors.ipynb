## Exercise 4 - Orthogonality
With the following vector we are going to calculate the orthogonality between them:
- u = [1, 2, 3]
- v = [-3, 0, 1]
- w = [2, -5, 4]

So, calculate the following operations `u·v`, `u·w` and `v·w` and determine which of them are orthogonal

````python
u = np.array([1, 2, 3])
v = np.array([-3, 0, 1])
w = np.array([2, -5, 4])
````

````python
u_v = u @ v
print(u_v) # 0 -> this are orthogonal
````

````python
u_w = u @ w
print(u_w) # > 0 acute angle
````

````python
v_w = v @ w
print(v_w) # < 0 obtuse angle
````

````python
angle = math.degrees(math.acos(u_v / np.linalg.norm(u) * np.linalg.norm(v)))
print(angle)
````

## Exercise 5 - Cosine similarity
Given the following embeddings simplified in three dimensions:
- cat = [0.8, 0.3, 0.1]
- dog = [0.7, 0.4, 0.2]
- car = [0.1, 0.0, 0.9]
- truck = [0.2, 0.1, 0.8]

Implement:
1. Create the function `cosine_similarity(a, b)` just with basic NumPy
2. Calculate the six similarities between four vectors
3. Order them ascending

````python
cat = np.array([0.8, 0.3, 0.1])
dog = np.array([0.7, 0.4, 0.2])
car = np.array([0.1, 0.0, 0.9])
truck = np.array([0.2, 0.1, 0.8])
````

````python
def cosine_similarity(a, b):
    return (a @ b) / (np.linalg.norm(a) * np.linalg.norm(b))
````

````python
cat_dog = cosine_similarity(cat, dog)
print(cat_dog)
cat_car = cosine_similarity(cat, car)
print(cat_car)
cat_truck = cosine_similarity(cat, truck)
print(cat_truck)
dog_car = cosine_similarity(dog, car)
print(dog_car)
dog_truck = cosine_similarity(dog, truck)
print(dog_truck)
car_truck = cosine_similarity(car, truck)
print(car_truck)
````

````python
similarities = [cat_dog, cat_car, cat_truck, dog_car, dog_truck, car_truck]
similarities.sort(reverse=True)
print(similarities)
````
