## Exercise 2 - Matrix multiplication
Given the following matrices:
``` python
A = [[1, 2, 3],
     [4, 5, 6]]
B = [[1, 0],
     [0, 1],
     [1, 0]]
```
1. Calculate A @ B
2. is it possible to calculate B @ A? If it is, what is its shape?
3. Create matrix identity of the right size and verify A @ I = A

````python
A = np.array([[1, 2, 3], [4, 5, 6]])
B = np.array([[1, 0], [0, 1], [1, 0]])
````

````python
PROD_AB = A @ B
print(PROD_AB)
````

````python
PROD_BA = B @ A
print(PROD_BA)
````

````python
print(f"A @ B shape: {PROD_AB.shape}")  # (2, 2)
print(f"B @ A shape: {PROD_BA.shape}")  # (3, 3)
# Conclusion: A@B != B@A — the matrices multiplication is not commutative
````

````python
# A shape (2x3), so identity has to be (3x3) for this valid result A @ I
# The rule: A (m x n) @ I (n x n) = A (m x n)
A_IDENTITY = np.eye(3)
NEW_A = A @ A_IDENTITY
print(NEW_A == A)
````

## Exercise 3: Transposed
Given de following matrix calculate:
``` python
A = [[1, 2, 3],
     [4, 5, 6]]
```
1. Transposed manually
2. Verify ```(A^T)^T = A``` with Numpy
3. Create a symmetric matrix 3x3 abd verify that ```S = S^T```
4. Given ```v = [1, 2, 3]``` calculate ```v^T @ v``` and ```v @ v^T``` with Numpy, what are the resulting shapes?, Did you recognize anyone?

````python
## 2) Verify (A^T)^T = A
A = np.array([[1, 2, 3], [4, 5, 6]])
print(np.allclose(A.T.T, A))
````

````python
## 3) Symmetric matrix
S = np.array([[4, 1, 5], [1, 6, 3], [5, 3, 9]])
print(np.allclose(S, S.T))
````

````python
## 4) v^T @ v vs v @ v^T
v = np.array([1, 2, 3])

# v @ v -> scalar 14  (shape: ())
# This is the v point product from itself = ||v||^2 = 1^2 + 2^2 + 3^2 = 14
print(v @ v)
print(f"shape: {(v @ v).shape}")

# v @ v^T -> matrix 3x3  (exterior product)
# Every element [i,j] = v[i] * v[j]
# This matrix appears in ML, in the sizes update (gradient outer product)
print(np.outer(v, v))
print(f"shape: {np.outer(v, v).shape}")
````
