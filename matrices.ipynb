## Exercise 1 - Basic operations
Given the following matrices:
``` python
A = [[1, 2],
     [3, 4]]
B = [[5, 6],
     [7, 8]]
c = 3
```
Make the following operations:
1. A + B
2. c * A
3. A^T
4. is A + B = B + A? Why?, explain

````python
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
c = 3
````

````python
SUM = A + B
print(SUM)
````

````python
SCALAR = c * A
print(SCALAR)
````

````python
AT = A.T
print(AT)
````

````python
SUMAB = A + B
SUMBA = B + A
print(SUMAB == SUMBA) # So yes, the sum is commutative
````
